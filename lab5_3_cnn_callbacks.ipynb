```python
import numpy as np
import tensorflow as tf
import keras
from keras import layers, models

np.random.seed(42)
tf.random.set_seed(42)

(x_train_raw, y_train_raw), (x_test_raw, y_test_raw) = keras.datasets.mnist.load_data()
x_train_cnn = (x_train_raw.astype('float32') / 255.0).reshape((-1, 28, 28, 1))
x_test_cnn = (x_test_raw.astype('float32') / 255.0).reshape((-1, 28, 28, 1))

class ConvCallback(keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        current_acc = logs.get('accuracy')
        if current_acc is not None and current_acc >= 0.995:
            print("\nДосягнуто 99,5% точності, тому навчання скасовується!")
            self.model.stop_training = True

callback = ConvCallback()

model_cnn1 = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model_cnn1.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model_cnn1.summary()

history = model_cnn1.fit(
    x_train_cnn, 
    y_train_raw, 
    epochs=10, 
    batch_size=128,
    callbacks=[callback],
    verbose=1
)
```
